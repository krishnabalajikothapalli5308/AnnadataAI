```python
import pandas as pd

df = pd.read_csv("Agriculture_price_dataset.csv")

# Keep only Potato records (enough for the model)
potato = df[df['Commodity'] == 'Potato']
print(f"Potato records: {len(potato)}")

# Save as sample — will be under 5MB
potato.to_csv("sample_dataset.csv", index=False)
print("Saved sample_dataset.csv")
```
